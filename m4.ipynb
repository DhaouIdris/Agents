```python
import importlib
from pathlib import Path

from exercises import ingestion

ingestion = importlib.reload(ingestion)
root = Path("corpus/passerelle")
if not root.is_dir():
    raise RuntimeError("Ouvrir le notebook depuis le dossier contenant pyproject.toml.")
manifest = ingestion.read_manifest(root)
RUN_EXERCISE = False  # Passer à True après avoir complété et enregistré load_document.

for meta in manifest:
    print(meta["doc_id"], meta["language"], meta["status"], meta["access_groups"])
print("Nombre de sources :", len(manifest))
```

## Exercice : load_document

1. Lire les octets avec `path.read_bytes()`.
2. Décoder ces octets en UTF-8, sans modifier espaces ni sauts de ligne.
3. Calculer leur SHA-256 hexadécimal.
4. Retourner `Document` avec les métadonnées, les groupes convertis en tuple,
   le texte et le hash. Conserver le chemin relatif du manifeste.

Le statut archivé et les groupes sont conservés. Leur présence ne constitue pas
un filtrage d'autorisation. Le helper de résolution du chemin est fourni.


```python
if RUN_EXERCISE:
    ingestion = importlib.reload(ingestion)
    documents = [ingestion.load_document(root, meta) for meta in manifest]
    first = documents[0]
    print(first.doc_id, first.language, first.version)
    print("Empreinte :", first.content_hash)
    print("Chemin relatif :", first.source_path)
    print("Nombre de documents chargés :", len(documents))
    print(first.text[:250])
else:
    print("Chargeur non exécuté : compléter load_document puis activer RUN_EXERCISE.")
```
